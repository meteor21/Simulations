## 5. Candidate registry and optional service history
The code checks the table schema before importing your existing database, avoiding the earlier missing-table problem. When no registry is available, the real cohort remains empty; synthetic names are never substituted into it.

Network downloading is optional. Official FEC registration data and congressional service histories have different purposes. The service-history import covers officeholders, not every losing candidate. A currently downloaded crosswalk is marked observed now, not backdated.

```python
def has_funding_schema(path):
    path = Path(path).expanduser()
    if not path.is_file():
        return False
    try:
        with closing(sqlite3.connect(path.resolve().as_uri()+"?mode=ro", uri=True)) as con:
            columns = {r[1] for r in con.execute("PRAGMA table_info(federal_candidate_cycle)")}
        return {"candidate_id", "candidate_name", "finance_cycle", "office", "state", "district", "party"} <= columns
    except sqlite3.Error:
        return False

paths = ([Path(SOURCE_DB_OVERRIDE)] if SOURCE_DB_OVERRIDE else []) + [
    PROJECT / "database/US_Election_Funding.sqlite",
    PROJECT / "database/US_Election_Simulator.sqlite",
    Path("/content/US_Election_Funding.sqlite"),
    Path("/content/US_Election_Simulator.sqlite"),
]
source_db = next((p for p in paths if has_funding_schema(p)), None)
if source_db:
    print("Imported registry rows:", import_funding(store, source_db, cycles=DOWNLOAD_CYCLES), "from", source_db)
else:
    print("No existing funding-schema database found. Explicit input records are still available.")

missing_cycles = [c for c in DOWNLOAD_CYCLES if not store.rows("SELECT 1 FROM cs_candidate_cycles WHERE cycle=? AND office IN ('H','S') LIMIT 1", (c,))]
if NETWORK_READY and DOWNLOAD_FEC_IF_MISSING and missing_cycles:
    display(download_fec_cycles(store, shared_client, missing_cycles, WORK / "downloads"))
if NETWORK_READY and DOWNLOAD_OFFICEHOLDERS:
    display(download_congress_history(store, shared_client, WORK / "downloads", min_year=1990))
if not RUN_NETWORK and (DOWNLOAD_FEC_IF_MISSING or DOWNLOAD_OFFICEHOLDERS):
    print("Download switches are set, but RUN_NETWORK=False: no download attempted.")
store.backup(CHECKPOINT)
```

## 9. Resolve sources, plan, and run a bounded collection batch
Set `RUN_NETWORK=True` to activate. The Media Cloud key is read without printing or saving it. Initial requests resolve national/local source IDs; remaining requests collect stories. Search jobs persist, including pagination. Rerun to resume within the same persistent 20-attempt ceiling; reruns do not reset it.

Source resolution is interleaved between relevant national and subnational sources. An unresolved panel is explicitly blocked or partial, not treated as zero news. Result caps and API quotas prevent an unlimited one-cell crawl.

```python
# Reviewed output remains the default; automatic annotations are handled in the next section.
reviewed_config = FeatureConfig(mode=MODE, reviewed_only=True)
run_report = None
if not NETWORK_READY:
    print("Network disabled. No live search, page fetch or API request was made.")
    if cohort:
        display(pd.DataFrame(workflow.collector.plan_campaign(cohort, WINDOW_START, AS_OF,
            provider=PROVIDER, dry_run=True, max_initial_jobs=MAX_INITIAL_JOBS)).drop(columns=["candidate_ids", "detail"], errors="ignore"))
elif len(cohort) != 24:
    print("Live pilot blocked: exactly 24 sourced congressional candidates are required.")
else:
    if FETCH_FULL_TEXT and importlib.util.find_spec("trafilatura") is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "trafilatura>=1.12,<3"])
    token = network_token if PROVIDER == "mediacloud" else None
    client = shared_client
    preview = workflow.collector.plan_campaign(cohort, WINDOW_START, AS_OF, provider=PROVIDER,
        dry_run=True, max_initial_jobs=MAX_INITIAL_JOBS)
    nationals = list(dict.fromkeys(d for p in preview if p["scope"] == "national" for d in p["detail"]["selected_domains"]))
    locals_ = list(dict.fromkeys(d for p in preview if p["scope"] == "subnational" for d in p["detail"]["selected_domains"]))
    from itertools import zip_longest
    priority_domains = [d for pair in zip_longest(nationals, locals_) for d in pair if d is not None]
    if PROVIDER == "mediacloud":
        display(workflow.sources.resolve(client, token, max_sources=min(MAX_SOURCE_RESOLUTIONS, 6), max_pages=1, domains=priority_domains))
    plan_report = workflow.collector.plan_campaign(cohort, WINDOW_START, AS_OF, provider=PROVIDER,
        run_label=f"congress-{CYCLE}-{WINDOW_START}-{AS_OF}", max_initial_jobs=MAX_INITIAL_JOBS)
    display(pd.DataFrame(plan_report).drop(columns=["candidate_ids", "detail"], errors="ignore"))
    run_report = workflow.run(cohort, AS_OF, run_network=True, provider=PROVIDER, token=token,
        client_override=client, fetch=FETCH_FULL_TEXT, feature_config=reviewed_config,
        budget=RunBudget(max_requests=MAX_HTTP_REQUESTS, max_search_jobs=MAX_SEARCH_JOBS,
                         max_fetch_articles=MAX_FETCH_ARTICLES, max_annotation_pairs=MAX_ANNOTATION_PAIRS),
        checkpoint_path=CHECKPOINT)
    print({k:v for k,v in run_report.items() if k not in {"features", "sources"}})
print("Persistent queue status:")
display(pd.DataFrame(workflow.collector.queue_status()))
print("Persistent total HTTP attempts:", ledger.used, "of", ledger.limit)
```

## 10. Optional local classifier and separate exploratory output
Set `RUN_NLP=True` to download the local model. CPU works for a small pilot; a Colab GPU can accelerate batches. The exact loaded revision is recorded. Model confidence is not a calibrated probability of correct annotation. Automatic outputs stay unreviewed and are not silently included in the reviewed series.

Headlines and text excerpts measure portrayal, not what a representative voter thinks. The reviewer must check attribution, ambiguous names and context. The original standing-index components are not fabricated when survey/pedigree data are missing.

```python
annotator = None
if RUN_NLP:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "transformers>=4.44,<5", "torch>=2.2", "sentencepiece>=0.2", "protobuf>=4"])
    if FETCH_FULL_TEXT:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "trafilatura>=1.12,<3"])
    import torch
    from charisma_lab.annotate import NLIAnnotator
    annotator = NLIAnnotator(device=0 if torch.cuda.is_available() else -1, batch_size=8)
    print("Annotation model/revision:", annotator.model_id)
    print(workflow.analyst.annotate(annotator, max_pairs=MAX_ANNOTATION_PAIRS))
reviewed_features = workflow.analyst.features(cohort, AS_OF, reviewed_config) if cohort else []
exploratory_features = workflow.analyst.features(cohort, AS_OF,
    FeatureConfig(mode=MODE, reviewed_only=False)) if cohort else []
def compact(rows):
    return pd.DataFrame([{"candidate_id":r["candidate_id"], "name":r["name"], "mode":r["mode"],
        "reviewed_only":r["reviewed_only"], "national":r["national"]["score"],
        "subnational":r["subnational"]["score"], "media_recency":r["media_recency_score"],
        "in_office":r["in_office_at_cutoff"], "flags":"; ".join(r["flags"])} for r in rows])
print("Reviewed evidence series:")
display(compact(reviewed_features))
print("Exploratory, potentially unvalidated automatic series:")
display(compact(exploratory_features))
store.backup(CHECKPOINT)
```
